```python
import numpy as np

# =============================================
# EXERCÍCIO 3 - BALANCEAMENTO DE SERVIDORES
# =============================================

np.random.seed(2026)

TAREFAS = np.array([
    12, 35, 40, 8, 15,
    22, 19, 45, 60, 31,
    14, 28, 50, 18, 25,
    33, 42, 10, 5, 29
])

NUM_SERVIDORES = 4
TAM_POP = 100
GERACOES = 200


# Calcula a carga de cada servidor
def calcular_cargas(individuo):

    cargas = np.zeros(NUM_SERVIDORES)

    for i, servidor in enumerate(individuo):
        cargas[servidor] += TAREFAS[i]

    return cargas


# Calcula o Makespan
def calcular_makespan(individuo):

    cargas = calcular_cargas(individuo)

    return max(cargas)


# Seleção por torneio
def selecionar(populacao):

    indices = np.random.choice(
        len(populacao),
        3,
        replace=False
    )

    melhor = min(
        indices,
        key=lambda i: calcular_makespan(populacao[i])
    )

    return populacao[melhor].copy()


# Cruzamento
def cruzar(pai1, pai2):

    ponto = np.random.randint(1, len(TAREFAS))

    filho = np.concatenate([
        pai1[:ponto],
        pai2[ponto:]
    ])

    return filho


# Mutação
def mutar(individuo):

    filho = individuo.copy()

    if np.random.rand() < 0.2:

        posicao = np.random.randint(
            0,
            len(TAREFAS)
        )

        filho[posicao] = np.random.randint(
            0,
            NUM_SERVIDORES
        )

    return filho


# Criação da população inicial
populacao = np.random.randint(
    0,
    NUM_SERVIDORES,
    size=(TAM_POP, len(TAREFAS))
)


# =============================================
# ALGORITMO GENÉTICO
# =============================================

for geracao in range(GERACOES):

    nova_populacao = []

    # Descobre o melhor indivíduo
    melhor = min(
        populacao,
        key=calcular_makespan
    )

    # Elitismo
    nova_populacao.append(melhor.copy())

    # Gera os novos indivíduos
    while len(nova_populacao) < TAM_POP:

        pai1 = selecionar(populacao)
        pai2 = selecionar(populacao)

        filho = cruzar(pai1, pai2)

        filho = mutar(filho)

        nova_populacao.append(filho)

    populacao = np.array(nova_populacao)


# =============================================
# RESULTADO FINAL
# =============================================

melhor = min(
    populacao,
    key=calcular_makespan
)

cargas = calcular_cargas(melhor)

makespan = calcular_makespan(melhor)


print("=" * 50)
print("EXERCÍCIO 3 - BALANCEAMENTO DE SERVIDORES")
print("=" * 50)

print("\nMelhor indivíduo:")
print(melhor)

print("\nCarga de cada servidor:")

for servidor in range(NUM_SERVIDORES):

    print(
        f"Servidor {servidor}: "
        f"{cargas[servidor]:.0f} segundos"
    )

print(
    f"\nMakespan final: "
    f"{makespan:.0f} segundos"
)


# Mostra as tarefas de cada servidor
print("\nDistribuição das tarefas:")

for servidor in range(NUM_SERVIDORES):

    print(f"\nServidor {servidor}:")

    for i in range(len(TAREFAS)):

        if melhor[i] == servidor:

            print(
                f"Tarefa {i} = "
                f"{TAREFAS[i]} segundos"
            )
```
