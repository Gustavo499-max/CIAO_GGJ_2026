```python
import numpy as np

# ==========================================
# EXERCÍCIO 2 - PENALIDADE POR SLA
# ==========================================

# Semente solicitada no exercício
np.random.seed(15)

# Matriz de latência com 6 nós
matriz_latencia = np.random.uniform(5, 80, (6, 6))

# Limite máximo permitido pelo SLA
LIMITE_SLA = 50

# Penalidade aplicada quando o SLA é ultrapassado
PENALIDADE = 1000


# Função para calcular o custo da rota
def calcular_custo_com_sla(rota, matriz):

    custo_total = 0
    penalidade_total = 0

    for i in range(len(rota) - 1):

        origem = rota[i]
        destino = rota[i + 1]

        latencia = matriz[origem, destino]

        # Soma a latência do enlace
        custo_total += latencia

        print(
            f"Enlace {origem} -> {destino}: "
            f"{latencia:.2f} ms"
        )

        # Verifica se o SLA foi ultrapassado
        if latencia > LIMITE_SLA:

            penalidade_total += PENALIDADE

            print(
                f"  SLA ultrapassado! "
                f"Penalidade de +{PENALIDADE} ms"
            )

    # Soma custo normal + penalidades
    return custo_total + penalidade_total


# Rota utilizada para teste
rota_teste = np.array([0, 1, 2, 3, 4, 5])


# Executa a função
custo_final = calcular_custo_com_sla(
    rota_teste,
    matriz_latencia
)


# Exibe o resultado
print("\n==============================")
print("RESULTADO DO EXERCÍCIO 2")
print("==============================")

print(f"Rota: {rota_teste}")

print(
    f"Custo Total com penalidades: "
    f"{custo_final:.2f} ms"
)
```
