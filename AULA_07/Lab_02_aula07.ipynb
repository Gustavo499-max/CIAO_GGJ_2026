```python
import numpy as np

# Para obter os mesmos resultados em cada execução
np.random.seed(42)

# ==========================================================
# PROBLEMA DA MOCHILA - BLINDAGEM DE ATIVOS
# ==========================================================

weights = np.array([12, 2, 1, 4, 1])   # Custo/Consumo de memória
values = np.array([4, 2, 1, 10, 2])    # Cobertura de risco/valor
max_weight = 15

# Parâmetros do Algoritmo Genético
pop_size = 10
num_genes = len(weights)
generations = 10
mutation_rate = 0.1

# ==========================================================
# INICIALIZAÇÃO DA POPULAÇÃO
# ==========================================================

# Cria uma população binária 10x5
population = np.random.randint(
    0, 2,
    size=(pop_size, num_genes)
)

# ==========================================================
# FUNÇÃO FITNESS
# ==========================================================

def calculate_fitness(ind):
    total_weight = np.sum(ind * weights)
    total_value = np.sum(ind * values)

    # Se ultrapassar o peso máximo,
    # o indivíduo recebe penalização
    if total_weight > max_weight:
        return 0

    return total_value


# ==========================================================
# SELEÇÃO POR TORNEIO
# ==========================================================

def tournament_selection(pop, fitnesses):

    # Seleciona dois indivíduos aleatoriamente
    indices = np.random.choice(
        len(pop),
        size=2,
        replace=False
    )

    ind1 = indices[0]
    ind2 = indices[1]

    # Retorna uma cópia do indivíduo
    # que possui o maior fitness
    if fitnesses[ind1] >= fitnesses[ind2]:
        return pop[ind1].copy()
    else:
        return pop[ind2].copy()


# ==========================================================
# CROSSOVER
# ==========================================================

def crossover(parent1, parent2):

    # Escolhe um ponto de corte
    point = np.random.randint(1, num_genes)

    child1 = np.concatenate([
        parent1[:point],
        parent2[point:]
    ])

    child2 = np.concatenate([
        parent2[:point],
        parent1[point:]
    ])

    return child1, child2


# ==========================================================
# MUTAÇÃO
# ==========================================================

def mutate(ind):

    for i in range(num_genes):

        if np.random.rand() < mutation_rate:

            # Inverte o bit
            # 0 -> 1
            # 1 -> 0
            ind[i] = 1 - ind[i]

    return ind


# ==========================================================
# LOOP EVOLUTIVO
# ==========================================================

for g in range(generations):

    # Calcula fitness de toda a população
    fitnesses = np.array([
        calculate_fitness(ind)
        for ind in population
    ])

    new_population = []

    # Gera uma nova população
    for _ in range(pop_size // 2):

        # Seleção dos pais
        p1 = tournament_selection(
            population,
            fitnesses
        )

        p2 = tournament_selection(
            population,
            fitnesses
        )

        # Crossover
        c1, c2 = crossover(p1, p2)

        # Mutação
        c1 = mutate(c1)
        c2 = mutate(c2)

        # Adiciona os filhos
        new_population.extend([c1, c2])

    # Atualiza a população
    population = np.array(new_population)

    # Fitness da nova população
    new_fitnesses = np.array([
        calculate_fitness(ind)
        for ind in population
    ])

    # Melhor indivíduo da geração
    best_index = np.argmax(new_fitnesses)

    print(
        f"Geração {g + 1}: "
        f"Melhor Fitness = {new_fitnesses[best_index]}"
    )


# ==========================================================
# RESULTADO FINAL
# ==========================================================

final_fitnesses = np.array([
    calculate_fitness(ind)
    for ind in population
])

best_index = np.argmax(final_fitnesses)

best_individual = population[best_index]

best_weight = np.sum(
    best_individual * weights
)

best_value = np.sum(
    best_individual * values
)

print("\n==============================")
print("        RESULTADO FINAL")
print("==============================")

print("Melhor indivíduo:", best_individual)
print("Peso total:", best_weight)
print("Valor total:", best_value)
print("Fitness:", final_fitnesses[best_index])

print("\n[LAB 02 - SUCESSO] Algoritmo Genético executado!")
```
