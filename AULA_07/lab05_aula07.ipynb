```python
import numpy as np

# Para garantir resultados reproduzíveis
np.random.seed(42)

# ==========================================================
# LAB 05 - HILL CLIMBING / BUSCA LOCAL
# FUNÇÃO RASTRIGIN
# ==========================================================

# Função Rastrigin
# Possui vários mínimos locais
# O mínimo global ocorre em x = [0, 0, ...]
def rastrigin(x):
    return 10 * len(x) + sum(
        x**2 - 10 * np.cos(2 * np.pi * x)
    )


# ==========================================================
# BUSCA LOCAL HILL CLIMBING
# ==========================================================

def local_search_hill_climbing(
    solution,
    step_size=0.01,
    max_steps=20
):
    """
    Mecanismo de Intensificação
    (Aprendizado Individual / Memética)
    """

    # Copia a solução inicial
    current_sol = np.copy(solution)

    # Calcula o fitness inicial
    current_fit = rastrigin(current_sol)

    # Executa a busca local
    for _ in range(max_steps):

        # ==================================================
        # TODO 1 - GERAR UM VIZINHO
        # ==================================================

        # Gera uma pequena alteração aleatória
        # em cada dimensão da solução
        neighbor = current_sol + np.random.uniform(
            -step_size,
            step_size,
            size=len(solution)
        )

        # Calcula o fitness do vizinho
        neighbor_fit = rastrigin(neighbor)

        # ==================================================
        # TODO 2 - VERIFICAR SE O VIZINHO É MELHOR
        # ==================================================

        # Como queremos minimizar a função,
        # um fitness menor representa uma solução melhor
        if neighbor_fit < current_fit:

            current_sol = neighbor
            current_fit = neighbor_fit

    # Retorna a melhor solução encontrada
    return current_sol, current_fit


# ==========================================================
# TESTE DA BUSCA LOCAL ISOLADA
# ==========================================================

initial_solution = np.array([
    2.5,
    -3.1
])

refined_solution, final_fit = (
    local_search_hill_climbing(initial_solution)
)


# ==========================================================
# RESULTADOS
# ==========================================================

print(
    f"[LAB 05] Solução Inicial: "
    f"{initial_solution} | "
    f"Fitness: {rastrigin(initial_solution):.4f}"
)

print(
    f"[LAB 05] Solução Refinada: "
    f"{refined_solution} | "
    f"Fitness: {final_fit:.4f}"
)


# Verificação da melhoria
improvement = (
    rastrigin(initial_solution) - final_fit
)

print(
    f"[LAB 05] Melhoria obtida: "
    f"{improvement:.4f}"
)
```
