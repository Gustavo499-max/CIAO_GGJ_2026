```python
import numpy as np

# ==========================================================
# LAB 04 - ATUALIZAÇÃO DE FEROMÔNIO NO ACO
# ==========================================================

# Grafo de Latência entre Roteadores (Matriz de Custo)
latency_matrix = np.array([
    [0, 5, 2, 9],
    [5, 0, 3, 1],
    [2, 3, 0, 7],
    [9, 1, 7, 0]
])

# Número de nós
num_nodes = len(latency_matrix)

# Matriz inicial de feromônio
pheromone = np.ones((num_nodes, num_nodes))

# Taxa de evaporação
rho = 0.25


# ==========================================================
# FUNÇÃO DE ATUALIZAÇÃO DO FEROMÔNIO
# ==========================================================

def update_pheromone(pheromone_matrix, paths, costs, rho):

    # ======================================================
    # TODO 1 - EVAPORAÇÃO
    # ======================================================

    # Reduz o feromônio existente de acordo com rho
    pheromone_matrix = (1 - rho) * pheromone_matrix

    # ======================================================
    # TODO 2 - DEPÓSITO DE FEROMÔNIO
    # ======================================================

    # Percorre cada caminho e seu respectivo custo
    for path, cost in zip(paths, costs):

        # Percorre todas as conexões do caminho
        for i in range(len(path) - 1):

            u = path[i]
            v = path[i + 1]

            # O depósito é inversamente proporcional ao custo
            pheromone_matrix[u][v] += (1.0 / cost)

    return pheromone_matrix


# ==========================================================
# SIMULAÇÃO MOCK PARA TESTE
# ==========================================================

mock_paths = [
    [0, 2, 1, 3],
    [0, 1, 3]
]

mock_costs = [
    6.0,
    6.0
]


# ==========================================================
# EXECUÇÃO
# ==========================================================

updated_pheromone = update_pheromone(
    pheromone,
    mock_paths,
    mock_costs,
    rho
)


# ==========================================================
# RESULTADO
# ==========================================================

print(
    "[LAB 04] Matriz de Feromônio Atualizada:\n",
    updated_pheromone
)
```
