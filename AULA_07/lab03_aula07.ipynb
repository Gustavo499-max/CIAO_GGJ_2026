```python
import numpy as np

# ==========================================================
# LAB 03 - PSO (Particle Swarm Optimization)
# ==========================================================

# Para obter os mesmos resultados em cada execução
np.random.seed(42)

# ==========================================================
# FUNÇÃO DE FITNESS
# ==========================================================

# Função de Custo Esférica
# Queremos encontrar o mínimo em x = 0 e y = 0
def fitness_function(position):
    return np.sum(position**2)


# ==========================================================
# PARÂMETROS DO PSO
# ==========================================================

num_particles = 10
dimensions = 2
max_iter = 15

w = 0.5   # Inércia
c1 = 1.5  # Componente Cognitiva (Individual)
c2 = 1.5  # Componente Social (Coletiva)


# ==========================================================
# INICIALIZAÇÃO DAS PARTÍCULAS
# ==========================================================

# Posições iniciais entre -5 e 5
X = np.random.uniform(
    -5,
    5,
    (num_particles, dimensions)
)

# Velocidades iniciais entre -1 e 1
V = np.random.uniform(
    -1,
    1,
    (num_particles, dimensions)
)


# ==========================================================
# MELHORES POSIÇÕES INICIAIS
# ==========================================================

# Melhor posição individual de cada partícula
pbest_X = np.copy(X)

# Fitness das melhores posições individuais
pbest_fitness = np.array([
    fitness_function(p)
    for p in pbest_X
])

# Encontra a melhor partícula do enxame
gbest_index = np.argmin(pbest_fitness)

# Melhor posição global
gbest_X = np.copy(
    pbest_X[gbest_index]
)


# ==========================================================
# LOOP PRINCIPAL DO PSO
# ==========================================================

for t in range(max_iter):

    for i in range(num_particles):

        # Valores aleatórios
        r1 = np.random.rand()
        r2 = np.random.rand()

        # ==================================================
        # ATUALIZAÇÃO DA VELOCIDADE
        # ==================================================

        V[i] = (
            (w * V[i])
            + (c1 * r1 * (pbest_X[i] - X[i]))
            + (c2 * r2 * (gbest_X - X[i]))
        )

        # ==================================================
        # ATUALIZAÇÃO DA POSIÇÃO
        # ==================================================

        X[i] = X[i] + V[i]

        # ==================================================
        # AVALIAÇÃO DO FITNESS
        # ==================================================

        current_fitness = fitness_function(X[i])

        # Verifica se encontrou uma posição individual melhor
        if current_fitness < pbest_fitness[i]:

            pbest_fitness[i] = current_fitness

            pbest_X[i] = np.copy(X[i])

            # Verifica se também é a melhor posição global
            if current_fitness < fitness_function(gbest_X):

                gbest_X = np.copy(X[i])

    # Mostra evolução a cada iteração
    print(
        f"Iteração {t + 1}: "
        f"Melhor Fitness = {fitness_function(gbest_X):.6f}"
    )


# ==========================================================
# RESULTADO FINAL
# ==========================================================

print("\n========================================")
print("          RESULTADO FINAL - PSO")
print("========================================")

print(
    f"[LAB 03] Melhor posição encontrada pelo Enxame (gbest): "
    f"{gbest_X}"
)

print(
    f"Melhor Fitness: "
    f"{fitness_function(gbest_X):.6f}"
)

print(
    f"Coordenada X: {gbest_X[0]:.6f}"
)

print(
    f"Coordenada Y: {gbest_X[1]:.6f}"
)
```
